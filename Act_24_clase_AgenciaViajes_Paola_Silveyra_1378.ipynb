## Clase AgenciaViajes

```python
class AgenciaViajes:
    def __init__(self, nombre, destinosOfrecidos, asesoresDisponibles, certificaciones):
        # 4 Atributos definidos
        self.nombre = nombre
        self.destinosOfrecidos = destinosOfrecidos
        self.asesoresDisponibles = asesoresDisponibles
        self.certificaciones = certificaciones

    # 3 Funciones (Métodos) operativas
    def cotizarPaquete(self, destino):
        return f"{self.nombre} está cotizando un paquete exclusivo a {destino}."

    def reservarVuelo(self, cliente):
        return f"{self.nombre} ha reservado un vuelo para {cliente}."

    def emitirSeguro(self, cliente):
        return f"{self.nombre} ha emitido un seguro de viaje para {cliente}."

# Creación de 2 objetos (Instancias)
agencia1 = AgenciaViajes("Viajes Global", ["Cancún", "París"], 5, ["IATA", "SECTUR"])
agencia2 = AgenciaViajes("Turismo Real", ["Tokio", "Roma"], 3, ["IATA"])

# Datos de ejemplo para los resultados
agencias = [agencia1, agencia2]

for ag in agencias:
    print(f"--- Agencia: {ag.nombre} ---")
    print(f"Certificaciones: {', '.join(ag.certificaciones)}")
    print(ag.cotizarPaquete("Madrid"))
    print(ag.reservarVuelo("Juan Pérez"))
    print(ag.emitirSeguro("Juan Pérez"))
    print("-" * 50)

    print("Programa terminado")
    print("Programa realizado por Paola Silveyra NL = 50")
```
